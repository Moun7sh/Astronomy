## 2. Part C: Theoretical ML Problem
### How do different resampling strategies alter the Bias-Variance Tradeoff in severely imbalanced photometric classification?

**The Theoretical Context:**
In imbalanced classification with high feature overlap (like photometry), a standard model exhibits **high bias** against the minority class (predicting everything as the majority) and **high variance** at the boundary (due to overlapping noise).

**1. Undersampling (Tomek Links - Member 1)**
- *Mechanism*: Removes majority boundary samples.
- *Tradeoff Effect*: Drastically **reduces boundary variance** (cleaner signal), but risks **increasing bias** if it removes valid representations of the majority class, creating an artificially easy boundary.

**2. Oversampling (Borderline-SMOTE - Member 2)**
- *Mechanism*: Synthesizes minority samples near the boundary.
- *Tradeoff Effect*: **Reduces minority class bias** (gives the model enough data to learn the quasar pattern), but risks **increasing variance** by hallucinating synthetic points in overlapping noise regions (overfitting to synthetic data).

**3. Hybrid Resampling (SMOTEENN - Member 3)**
- *Mechanism*: Adds minority samples, then deletes misclassified neighbors from *both* classes.
- *Tradeoff Effect*: Attempts to **minimize both bias and variance** simultaneously. However, it alters the fundamental data distribution the most, risking an overconfident model that fails on intrinsically ambiguous real-world test data.
